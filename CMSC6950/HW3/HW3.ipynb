### Question 2

````python
import math as m

def cot(x):
    if x == 0:
        print("Warning: cot(x) not defined for x=0")
    if abs(x) >= m.pi:
        print("Warning: Input might be out of range")
    
    assert -m.pi < x < m.pi, "x must be in the range (−pi, pi)."
    assert x != 0, "cot(x) is not defined for x=0."
        
    return m.cos(x)/m.sin(x)
````

### Question 3

````python
def csc(x):
    """
    Calculate csc(x) = 1/sin(x) and returns the value when well-defined.
    Prints error message when sin(x) = 0 or x is not a real number.
    """
    try:
        result = 1 / m.sin(x)
        return result
    except ZeroDivisionError:
        print("Error: csc(x) is undefined when sin(x) = 0")
        return None
    except TypeError:
        print("Error: Input must be a real number")
        return None
````

````python
# Test
csc(m.pi)
````

````python
# Test 3
csc("1j")
````

c. Compose the two decorators to write a function that evaluates the polynomial $p(x) = x^3 + 5x^2 −2x+ 3$. Hand in a transcript of your Jupyter notebook, showing your function definition and some tests that it produces the correct values. No points will be given for functions that evaluate $p(x)$ directly, only for functions that use the decorators.

````python
class add_constant(object):
    def __init__(self, constant):
        self.constant = constant
    def __call__(self, f):
        def result(x):
            return f(x) + self.constant
        return result

def x_times_f(f):
    def result(x):
        return x * f(x)
    return result

@add_constant(3)
@x_times_f
@add_constant(-2)
@x_times_f
@add_constant(5)
def p(x):
    return x
````

````python
# Test 1, we know if x = 0, p(0) = 3
p(0)
````

````python
# Test 2, if x = 1, we expect p(0) = 7
p(1)
````
