## Introduction to d3

The plotting and figure packages are good for variations on standard plotting techniques and some packaged plot and chart types; but if you want to do something original, sometimes the packages will not support that, and you need to turn to low-level coding. We're starting that now.

D3 is package that allows manipulation of basic drawing shapes on a web page. Although it also offers methods to build complex plots out of basic shapes, there is a learning curve to how you put all those pieces together. In the next set of materials we try to introduce and demonstrate some of the basic package capabilities to provide a starting point to explore d3 further.

There are a lot of packages built on top of D3: plotly, react-d3 https://www.npmjs.com/package/react-d3-library and C3 https://c3js.org/ are a few that come to mind. Some of these allow you to combine D3 code with higer level coding. Vega has a similar structure as plotly, with a different grammar that operates close to the D3 level of specification https://vega.github.io/vega-lite/

Recently, the D3 developers have built a high level plotting package based on their javascript notebooks (rather than python notebooks of jupyter) under the brand **Observable**. They not to get confused with the difference of using Observeable and using D3.js. These course notes will not cover use of Observable.

Unforunately for me, most of the examples on the D3js site are embedded in observable notebooks. My approach will be [using D3 in vanilla HTML](https://d3js.org/getting-started#d3-in-vanilla-html) since they are difficult to get working in jupyter notebooks (jupyter assumes you will not want to manipulate the underlying notebook code directly, so it is sometimes difficult to inject javascipt the way you want).

Note that d3 is a javascript package, so:
* it is principally used client-side to render directly to the html page
* if you want to mix d3 calls with plotly, you can try using the plotly js libraries https://plotly.com/javascript/ directly without python bindings. These will go on a web page with no need for jupyter or python.
* Working with client-side javascript does not work well in jupyter, you can get it to work with some struggle, but there will continue to be conflict bewteen the jupyter html code and the d3 javascript. It is much easer to use D3 with simple HTML coding.

## Some important prerequisite concepts

d3.js matches up html DOM document elements with data items from your data set; so when the data changes, the svg elements automatically update.

It helps -actually its necessary - to have some understanding of the DOM, so for those who need some intro here's some basic [HTML/DOM concepts](https://www.cs.mun.ca/~brown/courses/6934/notes/DOM.html). I've used these notes in other courses, so if you need a refresher they will help. The [w3schools tutorials](https://www.w3schools.com/js/js_htmldom.asp) are also thorough.

d3 is a javascript library that leans heavily on the concept of function chaining to express visualization structure in a semi-declarative form.  You'll need to be comfortable with function chaining and callbacks to be able to debug d3.js scripts.  

Here's a video (also from another course) that introduces simple javascript [data structures and function chaining](https://www.cs.mun.ca/~brown/courses/6934/notes/videos/JSreview.mp4) with a few simple examples, if you want to review those. Function chaining starts at 39:48 on the video.
