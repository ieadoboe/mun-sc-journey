#### Problem 1.2

I've copied two JSON data sets on US and China GDP from <https://github.com/jdorfman/awesome-json-datasets> which appear to have originated from World Bank data. In the cell below, I show how to read in the json data from the usa file into a python data structure. It's in a typical list-of-dictionary format.

Your problem is to add python code to extract the *date* and *value* data from the dictionaries into two lists, and put them in two arrays for the x and y axis so you can use matplotlib to plot the data on a line graph. 

```python
import json

# load an entire json structure from a file in one shot. Better not be a big data set!
with open('usa.json', 'r') as json_fp:
    usa_data = json.load(json_fp)

# this will print all the data with no layout:
print(usa_data[1])

# use the following for a pretty print of json data. I print the first 2 records..
print(json.dumps(usa_data[1][:2], indent=1))

# also consider using the debugger to look at the data structure. Ask the prof to demo this if you want.
```

```python
print(json.dumps(obj=usa_data[0], indent=1))
```

```python
# View first record
print(json.dumps(obj=usa_data[1][0], indent=1))
```

```python
# View second record for pattern
print(json.dumps(obj=usa_data[1][1], indent=1))
```

```python
import numpy as np

# Extract date and value into lists
dates = []
values = []

for record in usa_data[1]:  # usa_data[1] contains the actual data records
    dates.append(record['date'])
    values.append(record['value'])

# Convert to numpy arrays
usa_x = np.array(dates, dtype=int)  # Convert to integers for years
usa_y = np.array(values, dtype=float)  # Convert to floats for values

# Plot
plt.figure(figsize=(10, 6))
plt.plot(usa_x, usa_y, marker='o', linestyle='-', color='b', linewidth=2)
plt.xlabel('Year')
plt.ylabel('GDP (USD)')
plt.title('USA GDP Over Time')
plt.grid(True, alpha=0.3)
plt.show()
```

#### Problem 1.4

The examples given do not show how to put two lines on the same plot. As a extra challenge, research out how to add a second line to the plot and use the data from the china file to add the second line to a new plot. Add some cells for your new code and plots.


```python
import json

# Load China data
with open('china.json', 'r') as json_fp:
    china_data = json.load(json_fp)

# Extract date and value from China data
china_dates = []
china_values = []

for record in china_data[1]:
    china_dates.append(record['date'])
    china_values.append(record['value'])

# Convert to numpy arrays
china_x = np.array(china_dates, dtype=int)
china_y = np.array(china_values, dtype=float)

# Plot both USA and China on the same figure
plt.figure(figsize=(12, 7))
plt.plot(usa_x, usa_y, marker='o', markevery=5, color='b',
         linestyle='-', linewidth=2, label='USA GDP')
plt.plot(china_x, china_y, marker='s', markevery=5, color='r',
         linestyle='-', linewidth=2, label='China GDP')

plt.xlabel('Year')
plt.ylabel('GDP (USD)')
plt.title('USA vs China GDP Over Time')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
