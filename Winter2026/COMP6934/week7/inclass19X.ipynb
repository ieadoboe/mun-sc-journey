Consider the 3D line plot below, modified from problem 12.

Suppose that the purpose of the plot is to compare the temperature range (low to high) at various locations.


```python
import pandas as pd

# read the csv data into a dataframe
df = pd.read_csv('datasets/jash312i/NL_weather.csv')
df = df[df['Station Name'].str.contains("ST. JOHN'S INTL")]
df = df[df['Year'] == 2020]
df = df[['Station Name', 'Date/Time', 'Month', 'Max Temp (C)', 'Min Temp (C)']]
aggregated = df.groupby(by='Month').agg(
    {'Station Name': 'first', 'Max Temp (C)': 'max', 'Min Temp (C)': 'min'}).reset_index()
```

```python
# read the csv data into a dataframe
df = pd.read_csv('datasets/jash312i/ON_weather.csv')
df = df[df['Station Name'].str.contains("TORONTO INTL")]
df = df[df['Year'] == 2020]
df = df[['Station Name', 'Date/Time', 'Month', 'Max Temp (C)', 'Min Temp (C)']]
dfagg = df.groupby(by='Month').agg(
    {'Station Name': 'first', 'Max Temp (C)': 'max', 'Min Temp (C)': 'min'}).reset_index()
aggregated = pd.concat([aggregated, dfagg])
```

```python
# read the csv data into a dataframe
df = pd.read_csv('datasets/jash312i/BC_weather.csv')
df = df[df['Station Name'].str.contains("VANCOUVER INTL")]
df = df[df['Year'] == 2020]
df = df[['Station Name', 'Date/Time', 'Month', 'Max Temp (C)', 'Min Temp (C)']]
dfagg = df.groupby(by='Month').agg(
    {'Station Name': 'first', 'Max Temp (C)': 'max', 'Min Temp (C)': 'min'}).reset_index()
aggregated = pd.concat([aggregated, dfagg])
```

```python
import plotly.express as px
fig = px.line_3d(aggregated, x="Month", y="Min Temp (C)", z="Max Temp (C)",
                 width=900, height=600,
                 line_group='Station Name',
                 color='Station Name')
fig.show()
```

#### Problem 19

Describe and justify three reasons this is a poor choice of visualizaition.  At least two of your reasons should involve "rules of thumb" from Munzner chapter 6. You may use alternative design suggestions to support your critiaue.

* **No Unjustified 3D**
  
  3D introduces blockage, perspective distortion, and foreshortening, all of which reduce accuracy of perception. In this 3D plot, it may be difficult to understand what is going on here and gauge the actual change of time (here, months). 2D is almost always more readable.
* **Eyes beat memory**
  
  Working memory is severely limited. Spatial proximity allows preattentive, parallel comparison. Side by side views of the above plot is better than forcing the audience to learn the skill of correctly rotating, paning and zooming to be able to (1) view the plots correctly (2) understand the view they are looking at and (3) maintain the view in their working memory to compare.

* **Avoid overplotting**
