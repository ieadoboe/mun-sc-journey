Q1c. We define plex(n) to be a 1 followed by n zeros, so that plex(2) is 100, and plex(6) is
1 million. Write a python function that takes an integer, n as input, with default value 0,
prints a statement such as The plex of 3 is 1000. (correctly substituting the input
value, n, and plex(n)) and returns the value of plex(n). Run your code for 3 sample
values of n, at least one of which is a single digit, and at least one which is greater than
20.

```python
def plex(n=0):
    """
    Input:
        n: Integer (Default = 0)
        
    Returns:
        Plex of a given integer, 10^n
    """
    
    val = 10**n
    print(f"The plex of {n} is {val}")
    return val
    
plex()
    
```

```python
sample = (7, 15, 24)

for s in sample:
    plex(s)
```
