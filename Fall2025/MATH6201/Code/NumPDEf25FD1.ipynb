```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def fdapprox(x, u):
    """
    Reusable function to compute du/dx
    Take x, u such as u(x)
    Return du/dx
    Use 2nd order central internal grid points
    Use 2nd order one-sided at boundary grid points
    Modify the function as needed
    
    """
    n = u.shape[0]
    dudx = np.zeros_like(u)
    dx = x[1]-x[0]
    
    
    #left boundary
    dudx[0] = (-3*u[0]+4*u[1]-u[2])/(2*dx)

    #internal points
    for i in range(1,n-1):
        dudx[i] =(u[i+1]-u[i-1])/(2*dx)

    #right boundary
    dudx[n-1] = (3*u[n-1]-4*u[n-2]+u[n-3])/(2*dx)

    return dudx
```

```python
# create grid x and sample data u
n = 32
xmin = 0.0
xmax = 1.0
dx = (xmax-xmin)/(n-1)
x = np.linspace(xmin, xmax, n)

u = np.sin(2*np.pi*x)

# compute derivative
dx = x[1]-x[0]
dudx = np.zeros_like(u)
#left boundary
dudx[0] = (-3*u[0]+4*u[1]-u[2])/(2*dx)

#internal points
for i in range(1,n-1):
    dudx[i] =(u[i+1]-u[i-1])/(2*dx)

#right boundary
dudx[n-1] = (3*u[n-1]-4*u[n-2]+u[n-3])/(2*dx)

#exact solution
v = 2*np.pi*np.cos(2*np.pi*x)

# plot results
plt.plot(x,dudx,'b--')
plt.plot(x,v,'r-')
errInf = np.linalg.norm(dudx-v, ord=np.inf)
err = np.linalg.norm(dudx-v, ord=2)

print(errInf, err)
```
